```python
#carregar arquivo
import pandas as pd
import openpyxl

df = pd.read_excel("dados_ausentes.xlsx")
display(df)

```

```python
#Verificar se existem dados faltando

print(df.isnull())
```

```python
#contar numero de ausencias por colunas

print(df.isnull().sum())
```

```python
#Excluir a linha

df_sem_vazio = df.dropna()
print(df_sem_vazio)
```

```python
#Substituir ausencias
#Cidade por Desconhecido

df["Cidade"] = df["Cidade"].fillna("Desconhecido", inplace = False)
print(df)
```

```python
#Substituir dados ausentes
#Substituir salario por 0
df["Salário"] = df["Salário"].fillna(0.00, inplace = False)
print(df)

```

```python
df["Idade"] = pd.to_numeric(df["Idade"], errors='coerce')
df["Idade"] = df["Idade"].fillna(df["Idade"].mean(), inplace=False)

print(df)
df["Idade"] = df["Idade"].fillna(df["Idade"].mean(), inplace=False)


print(df)
```
