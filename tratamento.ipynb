```python
#carregar arquivo
import pandas as pd
import openpyxl

df = pd.read_excel("dados_ausentes.xlsx")
display(df)

```

```python
#Verificar se existem dados faltando

print(df.isnull())
```

```python
#contar numero de ausencias por colunas

print(df.isnull().sum())
```

```python
#Excluir a linha

df_sem_vazio = df.dropna()
print(df_sem_vazio)
```

```python
#Substituir ausencias
#Cidade por Desconhecido

df["Cidade"] = df["Cidade"].fillna("Desconhecido", inplace = False)
print(df)
```

```python
#Substituir dados ausentes
#Substituir salario por 0
df["Salário"] = df["Salário"].fillna(0.00, inplace = False)
print(df)

```

```python
#Criando média de idade
df["Idade"] = pd.to_numeric(df["Idade"])
df["Idade"] = df["Idade"].fillna(df["Idade"].mean(), inplace=False)
print(df)




```

```python
#erros de formatação (minusculos e maiusculo)

df["Nome"] = df["Nome"].str.capitalize()
df["Cidade"] = df["Cidade"].str.title()

print(df)
```

```python
#Padronização de nomenclatura

df["Cidade"] = df["Cidade"].str.upper().replace({
   "são paulo":"SP",
    "sao paulo": "SP",
    "SÃO PAULO":"SP",
    "rio de janeiro":"RJ",
    "RIO DE JANEIRO":"RJ"    
})

print(df)

```

```python
#Identificar tipos de dados em uma coluna
#Tipo str

coluna_string = df.select_dtypes(include=["string", "object"]).columns.to_list()
print(coluna_string)
```

```python
df.info()
```

```python
#Trocar o tipo de dado de idade para INT

df["Idade"] = df["Idade"].astype(int)
print(df)
```

```python
#Pegar o valor de cima para substituir ausencia

df["Idade"] = df["Idade"].ffill()

#Debaixo
df["Idade"] = df["Idade"].bfill()

```

```python
#criar nova coluna faixa-etaria

df["Faixa Etária"]=pd.cut(df["Idade"], bins=[0,18,35,60,100],labels=["Jovem", "Adulto", "Meia-Idade", "Idoso"])
print(df)
```

```python
#REMOVER DUPLICATAS

df_sem_duplicata = df.drop_duplicates(["Nome"])
print(df_sem_duplicata)
```

```python
#MONTAR CATEGORIA DE SALARIO

df["Faixa Salarial"] = df["Salário"].map(lambda x: "Baixo" if x < 1000 else("Médio" if x<3500 else "Alto"))
print(df)
```

```python
#Salvar alteracoes

df.to_excel("dados_tratados.xlsx", index=False)
```
